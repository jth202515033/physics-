```python
import math
```

수평으로 던진 공의 운동

```python
# 초기 속도 입력
v0 = float(input())
print("초기 속도 : ", v0,"m/s")
```

```python
# 초기 각도 입력
angle_degree = float(input())
print("초기 각도 : ", angle_degree,"도")

#각도를 radian으로 바꾸기
angle_radian = math.radians(angle_degree)
print("초기 각도 : ", angle_radian,"radian")
```

초기 속도

```python
#x방향의 초기 속도
vx = v0x = v0 * math.cos(angle_radian)
print("x방향의 초기 속도 : ", vx,"m/s")

#y방향의 초기 속도
vy = v0y = v0 * math.sin(angle_radian)
print("y방향의 초기 속도 : ", vy,"m/s")
```

중력가속도

```python
# 중력 가속도
g = 9.81
print(g)
```

최고점의 높이

```python
# 최고점의 높이
h = (v0y**2) * (math.sin(angle_radian)**2) / (2 * g)

print("최고점의 높이:", h, "m")
```

최고점에 도착한 시간

```python
#최고점에 도착한 시간
t1 = v0 * math.sin(angle_radian) / g
print("최고점에 도착한 시간 : ", t1, "s")
```

수평으로 이동한 최고 거리

```python
#수평으로 이동한 최고 거리
Rmax = v0 **2 / g
print("수평으로 이동한 최고 거리 : ", Rmax, "m")
```

지면에 닿을때의 Vx, Vy를 구하고 V를 벡터 성분 구하기

```python
vx = Rmax / t1
print("vx : ", vx, "m/s")

vy = g * t1
print("vy : ", vy, "m/s")

V = math.sqrt(vx**2 + vy**2)
print("V : ", V, "m/s")

angle_degree = math.degrees(math.atan2(vy, vx))
print(angle_degree, "도")
```

```python
import math
```

```python
g = 9.8
print("중력가속도 : ",g, "m/s")
```

```python
v0 = float(input())
print("초기속도 : ", v0, "m/s")
```

```python
y0 = float(input())
print("초기높이 : ", y0, "m")
```

```python
angle_degree = float(input())
angle_radian = math.radians(angle_degree)
print("각도 : ", angle_degree, "degree")
```

```python
y = 0
```

```python
v0x = v0 * math.cos(angle_radian)
v0y = v0 * math.sin(angle_radian)
```

a

```python
a = -1/2 * g
b = v0 * math.sin(angle_radian)
c = y0
t = (-b + math.sqrt(b**2 - 4*a*c)) / (2*a)
if t < 0:
  t = (-b - math.sqrt(b**2 - 4*a*c)) / (2*a)

print("시간: ", t, "초")
```

b

```python
dx = v0 * math.cos(angle_radian) * t
print("이동거리: ", dx, "m")
```

c

```python
vx = v0x
print("속도의 수평성분 : ", vx, "m/s")
```

```python
vy = v0y - g*t
print("속도의 수직성분 : ", vy, "m/s")
```

d

```python
V = math.sqrt(vx**2 + vy**2)
print("속도의 크기 :",V, "m/s")
```

e

```python
theta = math.atan2(vy, vx)
theta_degree = math.degrees(theta)
print("각도 :", theta_degree, "degree")
```

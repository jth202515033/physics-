```python
#include <math.h>
#include <stdio.h>

void main()
{
    int number;
    const double pi = 3.141592653589793;
    double x, y, r, theta_rad, theta_deg;

    printf("직교좌표에서 극좌표로 바꾼다면 1번, 극좌표에서 직교좌표로 바꾼다면 2번: ");
    scanf_s("%d", &number);

    if (number == 1)
    {
        printf("직교좌표의 x값을 입력하세요: ");
        scanf_s("%lf", &x);
        printf("직교좌표의 y값을 입력하세요: ");
        scanf_s("%lf", &y);

        r = sqrt(x * x + y * y);
        theta_rad = atan2(y, x);
        theta_deg = theta_rad * 180.0 / pi;

        printf("극좌표계: r = %lf, theta = %lf°\n", r, theta_deg);
    }
        if (number == 2)
    {
        printf("극좌표의 r값을 입력하세요: ");
        scanf_s("%lf", &r);
        printf("극좌표의 각도를 입력하세요 (도 단위): ");
        scanf_s("%lf", &theta_deg);

        theta_rad = theta_deg * (pi / 180.0);

        x = r * cos(theta_rad);
        y = r * sin(theta_rad);

        printf("직교좌표계: x = %lf, y = %lf\n", x, y);
    }
    else
    {
        printf("잘못된 입력입니다. 1 또는 2를 입력하세요.\n");
    }
}
```

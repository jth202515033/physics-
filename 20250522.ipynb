```python
import math

m = 0.15
r = 0.60
f = 2.0

#(a) 주기 T 계산
T = 1 / f
print("(a) 주기 T =",T,"s")

#(b) 속력 v 계산
v = 2 * math.pi * r / T
print("(b) 속력 v =", round (v,2),"m/s")

#(c) 구심가속도 ac 계산
ac = v**2 / r
print("(c) 가속도 ac =", round (ac,2),"m/s^2")
```

```python
import math

r = 5
g = 9.8
v = 10

# (1) theta 구하기 (T*sin(theta) = m*v**2 / r   ,    T*cos(theta) = mg)

tan_theta = v**2 / (r * g)

theta_radians = math.atan(tan_theta)

theta_degrees = math.degrees(theta_radians)

print("theta (도):", round(theta_degrees, 2))
```

```python
import math

r = 500
t = 10.5
v = 32

#(a) 접선 가속도 구하기
at = v / t
print(f"(a) 접선 가속도 at = {at:.2f} m/s^2")

#(b) 속력이 35m/s일때 지름 방향의 속도 구하기
vmax = 35
ar = vmax**2 / r
print(f"(b) 지름 가속도 ar = {ar:.2f} m/s^2")
```
